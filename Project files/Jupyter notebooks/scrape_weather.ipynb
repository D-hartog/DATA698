```python
# import neccessary libraries 
import pandas as pd
import regex as re
import numpy as np
import os
```

```python
# Read in xslx file 
weather_file = pd.read_csv('https://raw.githubusercontent.com/D-hartog/DATA698/refs/heads/main/Project%20files/Data/Weather_data%20-%20sheet1.csv')

weather_file.head()
```

```python
weather_file.replace("T", np.nan)
```

```python
# Replace any value "T" with NA

weather_file = weather_file.replace("T", np.nan)
weather_file.head()
```

```python
# View rows with NA value
weather_file[weather_file["Precipitation"].isna()]
```

```python
#T: Trace amount- Less than 0.01" precipitation; less than 0.1" snowfall; less than 1" snow depth.

# FIll in NA values based on above criteria 

weather_file["Precipitation"] =  weather_file["Precipitation"].fillna(0.009)
weather_file["Snowfall"] =  weather_file["Snowfall"].fillna(0.09)
weather_file["SnowDepth"] =  weather_file["SnowDepth"].fillna(0.9)
```
